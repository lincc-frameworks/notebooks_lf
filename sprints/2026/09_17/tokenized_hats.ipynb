## Tokenized Images & Spectra in HATS

```python
lsn_cat = lsdb.open_catalog("hf://datasets/brantd/mmu_ssl_legacysurvey_north_aion_tokens")
lsn_cat.plot_pixels()
lsn_cat
```

```python
desi_cat = lsdb.open_catalog("hf://datasets/brantd/mmu_desi_edr_sv3_aion_tokens")
desi_cat.plot_pixels()
desi_cat
```

```python
tokens_cat = lsn_cat.crossmatch(desi_cat, suffix_method="overlapping_columns")
tokens_cat.plot_pixels()
tokens_cat
```

```python
tokens_nf = tokens_cat.compute()
tokens_nf
```
